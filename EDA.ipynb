```python
# Calcular promedio de estrellas para cada atributo
attr_means = df[selected_cols].apply(lambda x: df.loc[x==1, "stars"].mean())
```

```python
# Top 10 atributos con mayor promedio de estrellas
top10_attrs = attr_means.sort_values(ascending=False).head(10)
print("Top 10 atributos asociados a mejores calificaciones:")
print(top10_attrs)
```

```python
# Crear matriz de correlación solo con top 10 atributos + stars
top_corr_matrix = df[top10_attrs.index.tolist() + ["stars"]].corr()

# Matriz de correlación
plt.figure(figsize=(10,8))
sns.heatmap(top_corr_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f",
            center=0,
            linewidths=0.5,
            mask=np.triu(np.ones_like(top_corr_matrix, dtype=bool)))
plt.title("Matriz de correlación: Top 10 atributos más asociados a calificaciones altas", fontsize=14)
plt.show()
```
