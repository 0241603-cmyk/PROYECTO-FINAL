```python
import pandas as pd
from google.colab import files
import plotly.graph_objects as go
import folium
from folium.plugins import BeautifyIcon



data_url2= "https://raw.githubusercontent.com/0241603-cmyk/PROYECTO-FINAL/refs/heads/main/hotels_ca.csv"
data_url3= "https://raw.githubusercontent.com/0241603-cmyk/PROYECTO-FINAL/refs/heads/main/Prof_BBDD_BI.csv"



df_coordenadas=pd.read_csv(data_url2)
df_profesor=pd.read_csv(data_url3)



#Eliminamos los que no están abiertos
df_coordenadas = df_coordenadas[df_coordenadas["is_open"] == 1]
df_coordenadas.reset_index(drop=True, inplace=True)

df_profesor = df_profesor[df_profesor["is_open"] == 1]
df_profesor.reset_index(drop=True, inplace=True)


#Nuevo df solo con columnas que necesito
df_coord=df_coordenadas[["name","latitude","longitude", "address"]]
df_prof=df_profesor[["name","latitude","longitude", "address"]]


#Juntamos los dos df
df_final=pd.concat([df_coord,df_prof])


#Checar duplicados
df_final["name_duplicada"] = df_final["name"].duplicated(keep=False)
print(df_final[df_final["name_duplicada"]])
df_final["address_duplicada"] = df_final["address"].duplicated(keep=False)
print(df_final[df_final["address_duplicada"]])
duplicados= df_final[(df_final["name_duplicada"] == True) &
    (df_final["address_duplicada"] == True)]
num_duplicados = len(duplicados)
print("Número de duplicados:", num_duplicados)
#Eliminamos duplicados
df_final = df_final.drop_duplicates(subset=["name", "address"], keep="first")
# Crear mapa centrado en California
mapa = folium.Map(location=[36.7783, -119.4179], zoom_start=6)

# Agregar hoteles como marcadores
for i, row in df_final.iterrows():
    if pd.notna(row["latitude"]) and pd.notna(row["longitude"]):
        folium.Marker(
            location=[row["latitude"], row["longitude"]],
            popup=f"{row['name']}<br>{row['address']}",  # Nombre y dirección al hacer clic
            #icon=folium.Icon(color="darkblue", icon="hotel", prefix="fa")
             icon=BeautifyIcon(
        icon="hotel",
        icon_shape="marker",
        background_color="darkblue",
        text_color="white",
        border_color="white",
        border_width=2)
        ).add_to(mapa)

# Guardar y mostrar mapa interactivo en Colab/Jupyter
mapa.save("mapa_hoteles_california.html")
from google.colab import files
files.download("mapa_hoteles_california.html")
```
