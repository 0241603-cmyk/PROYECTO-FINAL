```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.metrics.pairwise import cosine_similarity

import re
import string
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
data = "https://github.com/melody-10/Proyecto_Hoteles_California/blob/main/df_clean_reviews.csv?raw=true"
df = pd.read_csv(data, engine='python')
df.head()
```

```python
import gensim
from gensim import corpora

clean_corpus = df["clean_review"].apply(lambda x: x.split())

# Creating document-term matrix
dictionary = corpora.Dictionary(clean_corpus)
doc_term_matrix = [dictionary.doc2bow(doc) for doc in clean_corpus]
```

```python
from gensim.models import LdaModel

# LDA model
lda = LdaModel(doc_term_matrix, num_topics=9, id2word = dictionary)

# Results
print(lda.print_topics(num_topics=9, num_words=10))
```

```python
def print_topics(model, num_words=10):
    """
    Prints the topics and their top words from an LDA model.

    Args:
        model: The trained LDA model.
        num_words: The number of top words to print for each topic.
    """
    print("Top topics:")
    for topic_id, topic_words in model.print_topics(num_topics=-1, num_words=num_words):
        print(f"Tópico {topic_id}: {topic_words}")

print_topics(lda)
```

```python
# Get topic distributions for each document
topic_distributions = [lda.get_document_topics(doc) for doc in doc_term_matrix]

# Function to get the dominant topic for a document
def get_dominant_topic(topic_list):
    if not topic_list:
        return None
    # Find the topic with the highest probability
    return max(topic_list, key=lambda item: item[1])[0]

# Assign the dominant topic to each review
df["topic"] = [get_dominant_topic(topics) for topics in topic_distributions]

# Define topic labels
topic_labels = {
    0: "Services near Union Square",
    1: "Ammenities and spaces",
    2: "Great time and location",
    3: "Services",
    4: "Location and cleanliness",
    5: "Night life",
    6: "Staff",
    7: "Food and room service",
    8: "Pets"
}

# Map topic numbers to labels
df["topic_label"] = df["topic"].map(topic_labels)

```

```python
# Display the first few rows with the new columns
display(df[["name", "ratings", "text", "topic", "topic_label"]].head())
```
